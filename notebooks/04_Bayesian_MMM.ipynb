# 04 - Bayesian Marketing Mix Modeling (MMM)
We train a custom PyMC Bayesian Model incorporating Geometric Adstock Decay and Hill Saturation curves, using hierarchical priors for model stabilization.


```python
import pandas as pd
import pickle
import sys
sys.path.append('../src')

from preprocessing import load_merged_data
from mmm import fit_mmm

df = load_merged_data("../data")
res = fit_mmm(df, "mmm_results.pkl")
print("MMM Fit Complete.")
```

## Estimated ROIs by Marketing Channel


```python
rois_df = pd.DataFrame({
    'ROI': res['rois'],
    'Total Contribution': res['contributions']
})
rois_df.style.bar(subset=['ROI'], color='#58D68D')
```

## Marketing Channel Adstock Decay Rates (alphas)


```python
alphas_df = pd.DataFrame({'Alpha (Decay Rate)': res['alpha']})
alphas_df.style.bar(subset=['Alpha (Decay Rate)'], color='#5DADE2')
```

## Plot Saturation Curves (Diminishing Returns)


```python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(3, 2, figsize=(14, 12))
axes = axes.flatten()

for idx, c in enumerate(res['saturation_curves']):
    curve = res['saturation_curves'][c]
    axes[idx].plot(curve['spend'], curve['revenue'], color='#E74C3C', linewidth=2)
    axes[idx].set_title(f'Saturation Curve: {c}')
    axes[idx].set_xlabel('Daily Spend ($)')
    axes[idx].set_ylabel('Revenue Contribution ($)')
    axes[idx].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
