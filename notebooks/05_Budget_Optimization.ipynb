# 05 - Budget Allocation and Reallocation Optimization
We run the SciPy budget optimizer to find the optimal spends to maximize revenue. We also simulate a 55% spend reduction while preserving revenue.


```python
import pickle
import pandas as pd
import sys
sys.path.append('../src')

from optimizer import run_optimization_pipeline

res = run_optimization_pipeline("mmm_results.pkl", "optimization_results.pkl")
print("Optimization pipeline run complete.")
```

## Current Budget vs Optimized Allocation (Full Budget)


```python
comparison = pd.DataFrame({
    'Current Spend': res['current_avg_spends'],
    'Optimized Spend': res['opt_spends_current']
})
comparison['Difference'] = comparison['Optimized Spend'] - comparison['Current Spend']
print(f"Baseline Media Revenue: ${res['baseline_revenue_contribution']:,.2f}")
print(f"Optimized Media Revenue: ${res['opt_rev_current']:,.2f}")
comparison
```

## 55% Spend Saving Optimization Scenario


```python
saving_df = pd.DataFrame({
    'Current Spend': res['current_avg_spends'],
    'Reduced Optimized Spend': res['opt_spends_reduced']
})
saving_df['Reduction ($)'] = saving_df['Reduced Optimized Spend'] - saving_df['Current Spend']
print(f"Reduced Total Budget: ${res['reduced_total_budget']:,.2f} (55% Saving)")
print(f"Retained Revenue: ${res['opt_rev_reduced']:,.2f} ({res['revenue_retention_pct']:.2f}% of baseline media revenue!)")
saving_df
```

## Budget Reallocation Matrix Visualization


```python
import matplotlib.pyplot as plt
import numpy as np

labels = list(res['current_avg_spends'].keys())
current = list(res['current_avg_spends'].values())
opt_current = list(res['opt_spends_current'].values())
opt_reduced = list(res['opt_spends_reduced'].values())

x = np.arange(len(labels))
width = 0.25

fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(x - width, current, width, label='Current Allocation', color='#BDC3C7')
ax.bar(x, opt_current, width, label='Optimized Allocation (Same Budget)', color='#2ECC71')
ax.bar(x + width, opt_reduced, width, label='Optimized Allocation (Reduced Budget)', color='#E74C3C')

ax.set_ylabel('Daily Spend ($)')
ax.set_title('Budget Reallocation Comparisons')
ax.set_xticks(x)
ax.set_xticklabels(labels, rotation=30)
ax.legend()
plt.show()
```
