# 01 - Exploratory Data Analysis (EDA)
In this notebook, we load the raw datasets (sales, weather, holidays, discount, marketing) and explore trends, seasonality, correlation, and distribution of variables.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
sys.path.append('../src')

from preprocessing import generate_raw_data, load_merged_data

# Ensure data is generated
generate_raw_data("../data")
df = load_merged_data("../data")
print(f"Merged Dataset Shape: {df.shape}")
df.head()
```

## Summary Statistics


```python
df.describe()
```

## Historical Revenue Trend


```python
plt.figure(figsize=(12, 6))
plt.plot(df['Date'], df['Revenue'], label='Daily Revenue', color='#636EFA')
plt.title('Daily Revenue Trend (2 Years)', fontsize=14)
plt.xlabel('Date')
plt.ylabel('Revenue ($)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## Spend Distribution per Marketing Channel


```python
channels = ["TV Spend", "Facebook Spend", "Google Spend", "Instagram Spend", "Email Spend", "Affiliate Spend"]
spend_totals = df[channels].sum()

plt.figure(figsize=(10, 6))
sns.barplot(x=spend_totals.index, y=spend_totals.values, palette='viridis')
plt.title('Total Marketing Spend by Channel', fontsize=14)
plt.ylabel('Total Spend ($)')
plt.xticks(rotation=45)
plt.show()
```

## Correlation Heatmap


```python
numeric_cols = ['Revenue', 'Orders', 'Visitors', 'Temperature', 'Humidity', 'Rainfall'] + channels
plt.figure(figsize=(10, 8))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix', fontsize=14)
plt.show()
```
